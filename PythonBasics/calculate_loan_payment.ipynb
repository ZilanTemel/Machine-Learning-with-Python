```python
def calculate_loan(loan_amount, interest_rate, term):
    monthly_interest_rate = interest_rate / 100 / 12
    monthly_payment = (loan_amount * monthly_interest_rate * (1 + monthly_interest_rate) ** term) / \
                      ((1 + monthly_interest_rate) ** term - 1)
    return monthly_payment

loan_amount = float(input("Enter loan amount: "))
interest_rate = float(input("Enter annual interest rate (%): "))
term = int(input("Enter term (months): "))

monthly_payment = calculate_loan(loan_amount, interest_rate, term)
print(f"Your monthly payment: {monthly_payment:.2f}")
```
